# NumPy and Math

```python
import numpy as np
```

```python
# Essential: This links the nickname 'np' to the library

python_list = [1, 2, 3, 4, 5]

# Create a numpy array from a list
numpy_array = np.array(python_list)
print(numpy_array)
```

```python
oned_shape = numpy_array.shape
oned_size = numpy_array.size
oned_type = numpy_array.dtype

print('1D NumPy Array:')
print("Array:", numpy_array)
print(f'Shape: {oned_shape}')
print(f'Size: {oned_size}')
print(f'DataType: {oned_type}')
```

```python
arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([2, 4, 6, 8])
```

```python
# Addition performs element-wise math
addition = arr1 + arr2
print(addition)
```

```python
# Addition performs element-wise math
addition = arr1 + arr2
print(addition)
```

```python
subtraction = arr1 - arr2
print(subtraction)
```

```python
multiplication = arr1 * arr2
print(multiplication)
print(multiplication.dtype)
```

```python
# division = arr1 // arr2 # creates an int

division = arr1 / arr2 # creates an float
print(division)
print(division.dtype)
```

```python

data = np.array([85, 90, 78, 92, 88, 85, 95])

# Mean, Median
# Std, Variance, Range
data_mean = np.mean(data)
data_median = np.median(data)
data_std = np.std(data)
data_variance = np.var(data)
data_range = np.max(data) - np.min(data)

print(f'Mean: {data_mean}')
print(f'Median: {data_median}')
print(f'StDev: {data_std}')
print(f'Variance: {data_variance}')
print(f'Data Range: {data_range}')
```

```python
# Slicing and Indexing
sliced_array = numpy_array[0:3]

index_condition = numpy_array > 3 # creates an array of boolean values based on the condition

index_condition_array = numpy_array[numpy_array > 3] # grabbed (sliced) items from array based on condition

print('Sliced Array', sliced_array)
print('OG Array',numpy_array)
print('Array based on condition', index_condition)
print(index_condition_array)
```

```python
combined_result = numpy_array[1:4][numpy_array[1:4] > 3]
print('OG Array',numpy_array)
print('Combined Array:', combined_result)
```

```python
temp_data = np.array([22.3, 23.1, 24.5, 25.8, 23.6, 26.7, 27.9, 29.2, 30.5, 24.7, 23.4, 22.1, 25.3, 26.4, 28.7, 29.8, 31.2, 32.4, 30.7, 29.5, 27.8, 26.6, 23.9, 22.5, 24.1, 25.7, 27.3, 29.6, 31.0, 33.1, 31.9])
```

```python
mean_temp = np.mean(temp_data)
median_temp = np.median(temp_data)
std_dev_temp = np.std(temp_data)

print(f'Mean: {mean_temp}')
print(f'Median: {median_temp}')
print(f'Std Dev: {std_dev_temp}')
```

```python
temp_in_f = (temp_data * 9/5) + 32
print(temp_in_f)
```

```python
hot_days = temp_data[temp_data > 30]
print('Hot Temp Days:', hot_days)
print('Num of Hot Days:', len(hot_days))
```

```python
base_temp = 20
cooling_degrees_total = np.sum(np.maximum(temp_data - base_temp, 0))

print('Cooling Degrees:', cooling_degrees_total)
```
