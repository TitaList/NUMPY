```python
from numpy import random
```

```python
# random.rand() - creates random nums between 0 - 1
# random.randn() - generate a single random number
# random.randn(n) - generates an array of n random numbers
# random.randn(m, n) - generates a 2d array with m rows and n columns
# random.choice()
# random.shuffle() - shuffle contents of array
```

```python

# gives us array of ran numbers between 0 - 1
data = random.rand(10)

print(data)
```

```python
# Creates array of random numbers
# mean of 0
# strandard deviation of 1
random_numbers = random.randn(5)

print(random_numbers * 1000)
```

```python
elements = ['a', 'b', 'c', 'd', 'e', 'f']

# chooses a random element from array/list
random_sample = random.choice(elements)

print(random_sample)
```

```python
random.shuffle(elements)

print(elements)
```

```python
# Define a list of dishes on the menu
menu = [
    "Spaghetti Carbonara",
    "Chicken Alfredo",
    "Margherita Pizza",
    "Cheeseburger",
    "Caesar Salad",
    "Fish and Chips",
    "Pad Thai",
    "Sushi Platter",
    "Vegetable Stir-Fry",
    "Grilled Salmon"
]
```

```python
def suggest_dish():
    random_dish = random.choice(menu)
    return random_dish 
# returns ends a function AND allows us to return a value to the program
```

```python
print(suggest_dish())
```
